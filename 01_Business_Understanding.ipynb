### Fraud Behaviour
- Are frauds more common at night?
- Which merchant categories have higher fraud rates?
- Are online transactions riskier?
- Does a new device increase fraud probability?

### Risk Analysis
- Which features are most important?
- Can fraud be detected before payment is approved?
- Which ML model performs best?

### Business Questions for Financial Fraud Analysis

A bank can use transaction data to answer the following business questions:

- Which hour of the day has the highest number of fraudulent transactions?
- Which merchant categories (such as online shopping, travel, or entertainment) experience the most fraud?
- Does the transaction amount affect the probability of fraud?
- Which geographic locations or countries have the highest fraud rates?
- Are first-time transactions with a merchant more likely to be fraudulent than repeat transactions?
- Which payment methods (credit card, debit card, mobile wallet, or bank transfer) are most commonly associated with fraud?
- How does customer age or account history influence the likelihood of fraudulent transactions?
- Are transactions made during weekends or holidays more likely to be fraudulent than those on regular weekdays?
- How quickly do fraudsters perform multiple transactions after gaining access to an account?
- Which transaction patterns or customer behaviours are the strongest indicators of potential fraud before it occurs?
