### Features description

| Features | Data Type | Description | Category |
|:-----|:------:|:------:|:------:|
| step | Integer |Time step of the transaction (each step represents one hour in the simulation) |  Time |
| type | Object |Type of transaction | Categorical |
| amount | Integer |Amount involved in the transaction | Numerical |
| nameOrig | Object |Unique identifier of the customer initiating the transaction | Identifier |
| oldbalanceOrig | Float | Sender's account balance before the transaction | Numerical |
| newbalanceOrig | Float | Sender's account balance after the transaction | Numerical |
| nameDest | Object | Unique identifier of the receiving account | Identifier |
| oldbalanceDest | Float | Receiver's account balance before the transaction | Numerical |
| newbalanceDest | Float | Receiver's account balance before the transaction | Numerical |
| isFraud | Integer | Target variable indicating whether the transaction is fraudulent (1) or legitimate (0) | Target |
| isFlaggedFraud | Integer | Indicates whether the transaction was flagged by the bank's rule-based fraud detection system | Target |
